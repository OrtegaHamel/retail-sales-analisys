Carga de Datos

```python
import pandas as pd
import numpy as np

#Cargar el archivo en un Data Frame de Pandas
df = pd.read_csv('../data/retail_sales_dataset.csv')

#Mostrar las primeras 10 filas del Data Frame
df.head(10)
```

Exploración inicial

```python
# Mostramos las últimas 5 filas del Data Frame
print("=" * 50)
print("Últimas 5 filas: ")
display(df.tail(5))

# Mostramos información general del Data Frame
print("=" * 50)
print("Información general del Data Frame: ")
display(df.info())

# Generamos estadísticas descriptivas del Data Frame
print("=" * 50)
print("Estadísticas descriptivas del Data Frame: ")
display(df.describe())

# Mostramos el número de filas y columnas del Data Frame
print("=" * 50)
print("Número de filas y columnas del Data Frame: ")
display(df.shape)

# Mostramos si hay duplicados en el Data Frame
print("=" * 50)
print("Duplicados en el Data Frame: ")
display(df.duplicated().sum())

# Mostramos los valores nulos en cada columna del Data Frame
print("=" * 50)
print("Valores nulos en cada columna del Data Frame: ")
display(df.isnull().sum())
```

Transformación de Datos

```python
# Hacemos una copia del Data Frame original para trabajar con ella
df_retail = df.copy()

# Creamos la columna 'Customer_Segment' basada en la edad de los clientes
# Definimos los cortes de edad y sus etiquetas
bins_edad = [0, 25, 40, 60, 100]
etiquetas_edad = ['Gen Z', 'Millennial', 'Adulto', 'Senior']

df_retail['Customer_Segment'] = pd.cut(df_retail['Age'], bins=bins_edad, labels=etiquetas_edad)
# Mostramos las primeras 5 filas del Data Frame después de crear la columna 'Customer_Segment'
print("=" * 50)
print("Primeras 5 filas después de crear 'Customer_Segment': ")
display(df_retail.head(5))
```

```python

# Convertimos la columna 'Date' a formato datetime para normalizarla
df_retail['Date'] = pd.to_datetime(df_retail['Date'], format='%Y-%m-%d')

# Mostramos las primeras 5 filas del Data Frame después de la conversión
print("=" * 50)
print("Primeras 5 filas después de convertir 'Date' a datetime: ")
display(df_retail.head(5))
```

```python
# Crea una columna que clasifique las ventas en categorías significativas (‘Alta’, ‘Media’, ‘Baja’).
# Definimos los cortes de ventas y sus etiquetas
bins_ventas = [0, 100, 500, np.inf]
etiquetas_ventas = ['Baja', 'Media', 'Alta']
df_retail['Sales_Category'] = pd.cut(df_retail['Total Amount'], bins=bins_ventas, labels=etiquetas_ventas)
# Mostramos las primeras 5 filas del Data Frame después de crear la columna 'Sales_Category'
print("=" * 50)
print("Primeras 5 filas después de crear 'Sales_Category': ")
display(df_retail.head(5))
```

Agrupación y Agregación

```python
# Agrupamos por Categoría de Producto y Género
# Calculamos la suma, media y conteo de 'Total Amount' y la suma de 'Quantity' para cada grupo
producto_genero = df_retail.groupby(['Product Category', 'Gender']).agg({'Total Amount': ['sum', 'mean', 'count'], 'Quantity': sum})

# Mostramos el resultado del agrupamiento
print("=" * 50)
print("Resultado del agrupamiento por Categoría de Producto y Género:")
display(producto_genero)
```

```python
# Hacemos una agregación múltiple para obtener la suma, media y conteo, venta máxima y mínima, dispersión y varianza por Categoría de Producto
estadisticas_ventas = df_retail.groupby('Product Category')['Total Amount'].agg(['sum', 'mean', 'count', 'min', 'max', 'std', 'var'])
# Mostramos el resultado de la agregación múltiple
print("=" * 50)
print("Estadísticas de ventas por categoría de producto:")
display(estadisticas_ventas)
```

Análisis Personalizado con apply

```python
# Hacemos una función para clasificar el tipo de boleta según volumen y precio
def clasificar_boleta(fila):
    if fila['Quantity'] > 3 and fila['Price per Unit'] > 200:
        return 'Volumen Alto'
    elif fila['Quantity'] <= 2 and fila['Price per Unit'] > 400:
        return 'Compra Única Alta'
    elif fila['Total Amount'] < 100:
        return 'Low Cost'
    else:
        return 'Venta Estándar'

# Aplicamos la función fila por fila
df_retail['Ticket_Type'] = df_retail.apply(clasificar_boleta, axis=1)

print("Distribución de tipos de boleta:")
print(df_retail['Ticket_Type'].value_counts())
display(df_retail.head(5))
```

```python
# Calculamos el porcentaje de desviación de cada venta respecto a la media de su categoría de producto
# Primero calculamos la media de 'Total Amount' por categoría de producto
def calcular_desviación(fila):
    media_categoria = df_retail[df_retail['Product Category'] == fila['Product Category']]['Total Amount'].mean()
    desviacion = (fila['Total Amount'] - media_categoria) / media_categoria * 100
    return desviacion

# Aplicamos la función fila por fila
df_retail['Desviacion_Porcentual'] = df_retail.apply(calcular_desviación, axis=1)

# Mostramos las ventas que más se alejaron de la media de su categoría de producto
print("=" * 50)
print("Ventas que más se alejaron de la media de su categoría:")
display(df_retail[['Product Category', 'Total Amount', 'Desviacion_Porcentual']].sort_values(by='Desviacion_Porcentual', ascending=False).head(10))
```
