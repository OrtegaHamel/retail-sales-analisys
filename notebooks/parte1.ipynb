Carga y Procesamiento de Datos

```python
import pandas as pd
import numpy as np

# Carga de archivos
df = pd.read_csv('../data/retail_sales_dataset.csv')

# Comprobamos la carga
df.head()
```

```python
# Procesamiento de datos
print("Últimas 5 filas:")
display(df.tail()) #usamos display para mostrar el resultado de cada función
print("Información del DataFrame:")
display(df.info())
print("Estadísticas descriptivas:")
display(df.describe())
print("Tipos de datos:")
display(df.dtypes)
print("Número de filas y columnas:", df.shape)
```

Exploración de Datos con Numpy

```python
# Convertimos las colummnas a arrays de numpy
categorias_array = df['Product Category'].values
ventas_array = df['Total Amount'].values
fechas_array = df['Date'].values

# Valores únicos de categorías
categorias_unicas = np.unique(categorias_array)

# Calculamos días totales
dias_totales = np.unique(fechas_array).size

# Lista para almacenar resultados y econtrar max y min
lista_totales = []

# Exploramos por categoría
for cat in categorias_unicas:
    # Total de ventas por categoría
    total_ventas = np.sum(ventas_array[categorias_array == cat])
    lista_totales.append(total_ventas) #Guardamos el resultado en la lista
    
    # Promedio diario de ventas por categoría
    promedio_diario = total_ventas / dias_totales
    
    # Imprimimos resultados por categoría
    print(f"Categoría '{cat}': ")
    print(f"    Total de ventas: {total_ventas}")
    print(f"    Promedio diario: {promedio_diario:.2f}")
    print("-" * 50)

# Categorías de productos con mayores y menores ventas
max_ventas = np.max(lista_totales)
min_ventas = np.min(lista_totales)

# Usamos índices para encontrar las categorías correspondientes
categoria_max = categorias_unicas[np.argmax(lista_totales)]
categoria_min = categorias_unicas[np.argmin(lista_totales)]

# Imprimimos los resultados min y max
print(f"La categoría que más ventas tiene: {categoria_max} con {max_ventas}")
print(f"La categoría que menos ventas tiene: {categoria_min} con {min_ventas}")
```

Manipulación de Datos

```python
# Filtramos por categoría 'Electronics'
filtro_electronics = categorias_array == 'Electronics'
ventas_electronics = ventas_array[filtro_electronics]

#  Operaciones en los datos para obtener estadísticas adicionales
# Impuestos (Multiplicación)
iva = np.sum(ventas_electronics) * 0.19

# Monto neto (Resta)
monto_neto = np.sum(ventas_electronics) - iva

# Cargo por envío (Suma)
cargo_envio = np.sum(ventas_electronics) + 5.00

# Porcentaje del total de ventas (División)
total_ventas = np.sum(ventas_array)
total_ventas_electronics = np.sum(ventas_electronics)
porcentaje_ventas = (total_ventas_electronics / total_ventas) * 100

# Imprimimos resultados
print("Cálculos para la categoría 'Electronics':")
print(f" - Total de ventas: {np.sum(ventas_electronics)}")
print(f" - Impuestos: {iva:.2f}")  
print(f" - Monto neto: {monto_neto:.2f}")
print(f" - Cargos por envío: {cargo_envio}")
print(f" - Porcentaje del total de ventas: {porcentaje_ventas:.2f}%")
```
