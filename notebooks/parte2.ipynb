Carga de Datos

```python
import pandas as pd
import numpy as np

#Cargar el archivo en un Data Frame de Pandas
df = pd.read_csv('../data/retail_sales_dataset.csv')

#Mostrar las primeras 10 filas del Data Frame
df.head(10)
```

Exploración inicial de los Datos

```python
# Mostramos las últimas 5 filas del Data Frame
print("=" * 50)
print("Últimas 5 filas: ")
display(df.tail(5))

# Mostramos información general del Data Frame
print("=" * 50)
print("Información general del Data Frame: ")
display(df.info())

# Generamos estadísticas descriptivas del Data Frame
print("=" * 50)
print("Estadísticas descriptivas del Data Frame: ")
display(df.describe())

# Mostramos el número de filas y columnas del Data Frame
print("=" * 50)
print("Número de filas y columnas del Data Frame: ")
display(df.shape)

# Mostramos si hay duplicados en el Data Frame
print("=" * 50)
print("Duplicados en el Data Frame: ")
display(df.duplicated().sum())

# Mostramos los valores nulos en cada columna del Data Frame
print("=" * 50)
print("Valores nulos en cada columna del Data Frame: ")
display(df.isnull().sum())
```

Inspección de los Datos

```python
# Tipos de datos de cada columna del Data Frame
print("=" * 50)
print("Tipos de datos de cada columna del Data Frame: ")
display(df.dtypes)

# Valores únicos de la columna Producto
# Al no existir la columna 'Producto' en el Data Frame, se asume que se refiere a 'Product Category'
print("=" * 50)
print("Valores únicos de la columna 'Product Category': ")
display(df.value_counts('Product Category'))

# Muestra todos los valores únicos en la columna Tienda 
# Al no existir la columna 'Tienda' en el Data Frame, se decide inspeccionar 'Price per Unit'
print("=" * 50)
print("Valores únicos de la columna 'Price per Unit': ")
display(df['Price per Unit'].unique())
```

Filtrado de Datos

```python
# Filas donde las ventas (Ventas) sean mayores a 50
# Al no existir la columna 'Ventas' en el Data Frame, se decide filtrar por 'Total Amount'
print("=" * 50)
print("Filas donde las ventas son mayores a 50: ")
display(df[df['Total Amount'] > 50])

# Filas donde las ventas sean mayores a 500
# Como la mayoría las ventas están sobre 50, se decide mostrar también las filas donde las ventas sean mayores a 500 
print("=" * 50)
print("Filas donde las ventas son mayores a 500: ")
display(df[df['Total Amount'] > 500])

# Filas donde el precio (Precio) sea menor a 0.5
# Al no existir la columna 'Precio' en el Data Frame, se decide filtrar por 'Price per Unit'
print("=" * 50)
print("Filas donde el precio es menor a 0.5: ")
display(df[df['Price per Unit'] < 0.5])

# Filas donde el precio sea menor a 50
# Como no hay precios menores a 0.5, se decide mostrar también las filas donde el precio sea menor a 50
print("=" * 50)
print("Filas donde el precio es menor a 50:")
display(df[df['Price per Unit'] < 50])

# Filas donde el producto sea Manzanas y las ventas sean mayores a 30.
# Al no existir el producto 'Manzanas' en el Data Frame, se decide utilizar la categoría 'Electronics'
print("=" * 50)
print("Filas donde el producto es 'Electronics' y las ventas son mayores a 30: ")
display(df.query("`Product Category` == 'Electronics' and `Total Amount` > 30"))
```

Slicing de Datos

```python
# Mostramos sólo las columnas de Producto y Ventas
# Al no existir la columna 'Producto' en el Data Frame, se asume que se refiere a 'Product Category'
# Al no existir la columna 'Ventas' en el Data Frame, se asume que se refiere a 'Total Amount'
print("=" * 50)
print("Columnas Product Category y Total Amount:")
display(df[['Product Category', 'Total Amount']])

# Filas de las 5 a 10 y las columnas Producto y Tienda.
# Al no existir la columna 'Producto' en el Data Frame, se asume que se refiere a 'Product Category'
# Al no existir la columna 'Tienda' en el Data Frame, se decide filtrar por 'Price per Unit'
print("=" * 50)
print("Filas 5 a 10 de las columnas de Product Category y Price per Unit:")
display(df[['Product Category', 'Price per Unit']].iloc[5:11])

# Primeras 5 filas y primeras 3 columnas del Data Frame
print("=" * 50)
print("Primeras 5 filas y primeras 3 columnas del Data Frame:")
display(df.iloc[0:5, 0:3])
```
