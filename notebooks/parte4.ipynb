# Visualizaciones

```python
# Histogramas y boxplots para las variables numéricas
sns.set_style('whitegrid')

variables = [
    ('Age', 'skyblue',20),
    ('Price per Unit', 'lightgreen', 6),
    ('Quantity', 'salmon', 8),
    ('Total Amount', 'orchid', 20)]

for i, (col, color,n_bins) in enumerate(variables):
    fig, (ax_hist, ax_box) = plt.subplots(1, 2, figsize=(15, 5))

    sns.histplot(df[col], ax=ax_hist, kde=True, color=color, edgecolor='black', alpha=0.7, bins=n_bins)
    ax_hist.set_title(f'Histograma de {col}', fontweight='bold')
    ax_hist.set_xlabel(col)
    ax_hist.set_ylabel('Frecuencia')

    sns.boxplot(x=df[col], ax=ax_box, color=color, fliersize=8)
    ax_box.set_title(f'Boxplot de {col}', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Edad
Distribución uniforme. Los clientes de 20, 40 y 60 años se observan en proporciones similares. El bloxpot se ve centrado, sin sesgo generacional.
## Precio por Unidad
Se detecta un volumen de ventas uniforme en los segmentos de precio medio y alto, lo que sugiere que los clientes de gama media son tan frecuentes como los clientes de gama alta.
## Cantidad
Las barras se observan de tamaño similar con brechas similares entre ellas, por lo que el volumen de compra es constante y predecible, sin valores atípicos.
## Monto total
Presenta un sesgo positivo. Se observa que la mayor parte de las ventas ocurren por debajo de los $500, aunque se observan algunas compras extremas de $2000.

```python
# Convertimos la columna 'Date' al formato datetime oficial de Pandas
df['Date'] = pd.to_datetime(df['Date'])

plt.figure(figsize=(12, 6))

# Agrupamos el total de ventas por mes
sales_trend = df.groupby(df['Date'].dt.to_period('M'))['Total Amount'].sum().reset_index()
sales_trend['Date'] = sales_trend['Date'].dt.to_timestamp() # Convertimos el período a datetime

sns.lineplot(data=sales_trend, x='Date', y='Total Amount', 
             marker='o', color="#c50f27", linewidth=2.5)

plt.title('Tendencia Global de Ventas Mensuales (2023)', fontsize=14, fontweight='bold')
plt.xlabel('Mes')
plt.ylabel('Ventas Totales ($)')
plt.grid(True, alpha=0.6)

plt.tight_layout()
plt.show()
```

Se observa la forma de dientes de sierra, que indica un comportamiento de alta volatilidad con picos y valles marcados.  
El crecimiento más acelerado se observa en tre septiembre y octubre, de lo que se puede inferir que se realizó alguna acción para detener la caída.  


```python
sns.set_style('whitegrid')

# Gráfico de dispersión entre 'Age' y 'Total Amount'
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x='Price per Unit', y='Total Amount', 
                hue='Quantity', palette='viridis', s=100, alpha=0.8, edgecolor='black')
plt.title('Relación entre Precio y Monto Total de Compra', fontsize=14, fontweight='bold')
plt.xlabel('Precio por Unidad')
plt.ylabel('Monto Total de Compra')
plt.legend(title='Unidades por compra', loc='upper left')

plt.tight_layout()
plt.show()

```

Podemos ver que en la zona de productos de bajo costo hay alta densidad de puntos, donde ocurre el mayor volumen de transacciones.  
También vemos una cantidad importante de puntos en la zona de los precios altos, desde 300.  
El vacío entre los 100 y 300 sugiere que no hay productos de gama media, sólo precios extremos.  
La estrategia de la tienda debería enfocarse en tratar de llenar el inventario de productos de gama media.

```python
sns.set_style('whitegrid')

# Creamos una figura con dos subplots (compartiendo el eje X)
fig, (ax_box, ax_hist) = plt.subplots(2, sharex=True, 
                                      gridspec_kw={"height_ratios": (.15, .85)}, 
                                      figsize=(12, 8))

# 2. Boxplot (en la parte superior para identificar outliers)
sns.boxplot(x=df['Total Amount'], ax=ax_box, color='orchid', fliersize=8)
ax_box.set(title='Detección de Outliers y Distribución de Ventas', xlabel='')

# 3. Histograma (en la parte inferior para ver la densidad)
sns.histplot(df['Total Amount'], ax=ax_hist, kde=True, color='orchid', bins=20)
ax_hist.set(xlabel='Monto Total de Venta ($)', ylabel='Frecuencia')

# Añadimos líneas para marcar la media y la mediana en el histograma
plt.axvline(df['Total Amount'].mean(), color='red', linestyle='--', label=f"Media: {df['Total Amount'].mean():.2f}")
plt.axvline(df['Total Amount'].median(), color='blue', linestyle='-', label=f"Mediana: {df['Total Amount'].median():.2f}")
plt.legend()

plt.tight_layout()
plt.show()
```

La distribución de ingresos presenta un sesgo positivo. La presencia de outliers significativos por encima de los $1,500 infla el promedio general.  
Se puede observar que la media está siendo arrastrada hacia la derecha por los outliers, mientras la mediana nos entrega una visión más robusta respecto del total de ventas.

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Product Category', y='Total Amount', 
            hue='Product Category', palette='viridis', legend=False)

plt.title('Detección de Outliers por Categoría de Producto', fontsize=15, fontweight='bold')
plt.xlabel('Categoría de Producto')
plt.ylabel('Monto Total de Venta ($)')

plt.show()

# 3. Creación de Histogramas por Categoría (Distribución)
g = sns.FacetGrid(df, col="Product Category", hue="Product Category", 
                  palette='viridis', height=5, aspect=1.2)

g.map(sns.histplot, "Total Amount", kde=True, bins=15, alpha=0.7)

g.set_axis_labels("Monto Total ($)", "Frecuencia")
g.set_titles("{col_name}", fontweight='bold')

plt.tight_layout()
plt.show()
```

Las cajas de las tres categorías son similares en altura y posición. Esto indica que la estrategia de precios y el volumen de compra están estandarizados en toda la tienda.  
Se oberva un sesgo a la derecha. La mayoría de las ventas se concentran en el rango de $0 a $500, pero se extienden hasta los $2,000. Esto puede ser debido a que la tienda tiene un límite de gasto máximo por transacción autoimpuesto por su catálogo.  
Aparecen valores atípicos por encima del bigote superior de la columna Clothes, que representan transacciones excepcionales a investigar. 

```python
# Filtramos para ver exactamente qué pasó con esos outliers de Clothing
outliers_ropa = df[(df['Product Category'] == 'Clothing') & (df['Total Amount'] > 1500)]
print(outliers_ropa)
```
